```python
# Retículo en una recta (pseudo trivial)
v = vector([1, 2])
w = vector([3,0])
var('t')

points = [i*v + j*w for i in range(-1,2) for j in range(-1,2)]

p = list_plot(points, size=30, color='blue', axes = False)
for k in range(-1,2):
    p += parametric_plot((t*v[0] + k*w[0], t*v[1] + k*w[1]),
                         (t, -5, 5),
                         linestyle=':', color='gray', alpha=0.5)
    p += parametric_plot((t*w[0] + k*v[0], t*w[1] + k*v[1]),
                         (t, -5, 5),
                         linestyle=':', color='gray', alpha=0.5)


p += point((0,0), size=30, color='red')
p+= line([(0,0.05), (v[0], v[1])], alpha=0.5, thickness =2)
p+= line([(0.05,0), (w[0], w[1])], alpha=0.5, thickness =2)

p += text(r'$w_1$', (v[0] -0.5, v[1]+0.3), fontsize = 20, color = 'black', clip = True)
p += text(r'$w_2$', (w[0] +0.4, w[1]-0.3), fontsize = 20, color = 'black', clip = True)


#p += point((0,0), size=30, color='red')
#p += parametric_plot((t*v[0], t*v[1]), (t, -5, 5),
#                linestyle=':', ymin=-3, ymax=3, color = 'gray', alpha = 0.7)

p.show(aspect_ratio=1, gridlines=True, xmin = -5, xmax = 5, ymin = -3, ymax = 3)
p.save("ret.pdf")
```
